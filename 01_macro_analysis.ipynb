```python
from pyspark.sql.functions import (
    col, count, when, to_date, try_to_date, year, explode, split, regexp_replace, regexp_extract, sum, when
)
```

```python
data = spark.table("workspace.default.steam_game_output")
data = data.toDF(*[c.replace(".", "_") for c in data.columns])
display(data.limit(10))
```

# General Analysis of data distribution

## Which publisher has released the most games on Steam?

```python
top_publishers = (
    data
    .groupBy("publisher")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)

display(top_publishers)
```

#### Big Fish Games had published more games than any other name of the industry listed in this dataset.

```python
top_publishers.write.mode("overwrite").saveAsTable("top_publisher")
```

## What are the best rated games?


```python
top_games = (
    data
    .select("name", "positive")
    .orderBy(col("positive").desc())
    .limit(10)
)

display(top_games)
```

#### The best rated games were Counter Strike: Global Offensive; Dota 2, and Grand Theft Auto 5

```python
top_games.write.mode("overwrite").saveAsTable("top_games")
```

## Are there years with more releases? Were there more or fewer game releases during the Covid, for example?

```python
df_years = (
    data
    .withColumn("release_date", try_to_date(col("release_date"), "yyyy/M/d"))
    .withColumn("release_year", year(col("release_date")))
    .groupBy("release_year")
    .count()
    .orderBy("release_year")
)

display(df_years)
```

```python
df_years_clean = (
    data
    .withColumn("release_date_clean", try_to_date(col("release_date"), "yyyy/M/d"))
    .filter(col("release_date_clean").isNotNull())
    .withColumn("release_year", year(col("release_date_clean")))
    .groupBy("release_year")
    .count()
    .orderBy("release_year")
)

display(df_years_clean)
```

#### most releases happened after 2015 and peaked during the covid in 2020-2021

```python
df_years_clean.write.mode("overwrite").saveAsTable("df_years")
```

## How are the prizes distributed? Are there many games with a discount?


```python
from pyspark.sql.functions import round
df_price = (
    data
    .withColumn("price", col("price").cast("double"))
    .filter(col("price") > 1)
)
df_price = df_price.withColumn("price", (col("price") /100 ))
df_price = df_price.select("price")
display(df_price)
```

```python
df_price.write.mode("overwrite").saveAsTable("df_price")
```

```python
df_discount = (
    data
    .withColumn("has_discount", when(col("discount") > 0, 'yes').otherwise('no'))
)
df_discount = df_discount.select("appid", "has_discount")
display(
    df_discount.groupBy("has_discount").count()
)
```

#### After visualization it was found that very few discounts were in this dataset.

```python
df_discount.write.mode("overwrite").saveAsTable("df_discount")
```

## What are the most represented languages?


```python
df_lang = (
    data
    .select(explode(split(col("languages"), ";")).alias("language"))
    .groupBy("language")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)

display(df_lang)
```

#### English is used widely and internationally, it is of not surprise that we find it overwhelmingly available in games.

```python
df_lang.write.mode("overwrite").saveAsTable("df_lang")
```

## Are there many games prohibited for children under 16/18?

```python
df_age = (
    data
    .groupBy("required_age")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)

display(df_age)
```

```python

df_clean = data.withColumn(
    "required_age_clean",
    regexp_extract(col("required_age"), r"(\d+)", 1).cast("int")
)
```

```python
df_age2 = df_clean.withColumn(
    "age_group",
    when(col("required_age_clean") == 0, "No restriction (0)")
    .when((col("required_age_clean") >= 16) & (col("required_age_clean") < 18), "16-17")
    .when(col("required_age_clean") >= 18, "18+")
    .otherwise("Unknown")
)

display(df_age2.groupBy("age_group").count())
```

```python
age_dist = df_age2.groupBy("age_group").count().orderBy("count", ascending=False)
total = df_age2.count()

age_prop = (
    age_dist
    .withColumn("percentage", col("count") / total * 100)
)

display(age_prop)
```

#### Only ~.54 % of games were made for youth between 16 and 18 years old while ~98.81 % had no age restriction.
#### It sounds economically benefitial for a company to release their  games for any age as it widens the audience

```python
age_prop.write.mode("overwrite").saveAsTable("age_prop")
```
